```python
%%writefile my.py

def drawSpiral(myTurtle, maxSide):
    current_angle = 90
    for sideLength in range(1, maxSide + 1, 5):
        myTurtle.forward(sideLength)
        myTurtle.right(current_angle)
        current_angle -= 1
```

```python
import turtle
from my import *
t = turtle.Turtle()
drawSpiral(t,100)
```
